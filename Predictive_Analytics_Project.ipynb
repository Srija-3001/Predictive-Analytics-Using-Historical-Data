```python
# Predictive Analytics Using Historical Data

import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression

# Load Dataset
data = pd.read_csv('sales_forecast_data.csv')

# Display Dataset
print("Historical Sales Data:")
print(data)

# Data Preprocessing
data['Month_Number'] = range(1, len(data) + 1)

X = data[['Month_Number']]
y = data['Sales']

# Train Linear Regression Model
model = LinearRegression()
model.fit(X, y)

# Predict Future Sales
future_months = pd.DataFrame({
    'Month_Number': [13, 14, 15, 16]
})

predictions = model.predict(future_months)

# Print Predictions
print("\nPredicted Future Sales:")
for month, sale in zip(
    ['Jan Next Year', 'Feb Next Year', 'Mar Next Year', 'Apr Next Year'],
    predictions
):
    print(f"{month}: {sale:.2f}")

# Visualization
plt.figure(figsize=(10, 6))

# Historical Data
plt.scatter(
    data['Month_Number'],
    data['Sales'],
    label='Historical Sales'
)

# Regression Line
plt.plot(
    data['Month_Number'],
    model.predict(X),
    label='Trend Line'
)

# Future Predictions
plt.scatter(
    future_months['Month_Number'],
    predictions,
    label='Predicted Sales'
)

plt.xlabel('Month Number')
plt.ylabel('Sales')
plt.title('Sales Forecasting Using Linear Regression')
plt.legend()
plt.grid(True)

plt.show()
```
